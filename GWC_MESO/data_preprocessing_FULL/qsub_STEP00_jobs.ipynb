## Dscale baseline

````python
IND_max = 46000
INDs = np.arange(0, IND_max+1000, 1000)
INDs[-1] = IND_max
````

````python
for i, ind_start in enumerate(INDs[:-1]):
    
    ind_end = INDs[i+1]
    
    f = open('{}gather_prog_dscale_{:03d}.sh'.format(conf['qsub']['qsub_loc'], i), 'w') 
    
    heads = '''#!/bin/bash -l

#PBS -N gather_prog
#PBS -A P48500028
#PBS -l walltime=23:59:59
#PBS -l select=1:ncpus=12:mem=128GB
#PBS -q casper
#PBS -o gather_prog.log
#PBS -e gather_prog.err

conda activate credit
cd {}
python gather_prog_dscale_base.py {} {}
'''.format(conf['qsub']['scripts_loc'], ind_start, ind_end, ind_start, ind_end)
    
    print(heads, file=f)    
    f.close()

f = open('{}step00_gather_prog_dscale_all.sh'.format(conf['qsub']['qsub_loc']), 'w')

for i, ind_start in enumerate(INDs[:-1]):
    print('qsub gather_prog_dscale_{:03d}.sh'.format(i), file=f)
    
f.close()
````

````python
years = np.arange(2020, 2025)
````

````python
for i_year, year in enumerate(years):
    
    f = open('{}diag_dscale_{}.sh'.format(conf['qsub']['qsub_loc'], year), 'w') 
    
    heads = '''#!/bin/bash -l

#PBS -N {}_dscale
#PBS -A P48500028
#PBS -l walltime=23:59:59
#PBS -l select=1:ncpus=12:mem=256GB
#PBS -q casper
#PBS -o {}.log
#PBS -e {}.err

conda activate credit
cd {}
python merge_diag_dscale_base.py {}
'''.format(year, year, year, 
           conf['qsub']['scripts_loc'], 
           year)
    
    print(heads, file=f)
    f.close()

f = open('{}batch_diag_dscale.sh'.format(conf['qsub']['qsub_loc']), 'w')
for i_year, year in enumerate(years):
    print('qsub diag_dscale_{}.sh'.format(year), file=f)
f.close()
````
